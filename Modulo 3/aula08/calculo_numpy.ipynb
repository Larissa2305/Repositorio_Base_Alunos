```python
import numpy as np
import time


ARRAY_SIZE = 50_000_000
array_a = np.random.rand(ARRAY_SIZE)
array_b = np.random.rand(ARRAY_SIZE)

# Abordagem Vetorizada com NumPy
# Agora, vamos fazer o mesmo cálculo usando as operações vetorizadas do NumPy.
# Elas operam sobre o array inteiro de uma vez, sem a necessidade de um loop explícito.
def sum_of_squares_numpy_vectorized(a, b):
    return np.sum(a**2 + b**2)

# Medir o tempo de execução
start_time = time.time()
sum_np = sum_of_squares_numpy_vectorized(array_a, array_b)
end_time = time.time()

time_np = end_time - start_time
print("\nEtapa 4: Cálculo com NumPy vetorizado concluído.")
print(f"Resultado do NumPy: {sum_np:.4f}")
print(f"Tempo de execução (NumPy vetorizado): {time_np:.4f} segundos")
```
