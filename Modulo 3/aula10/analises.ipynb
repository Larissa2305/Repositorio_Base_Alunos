```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('Compras.csv', sep=';')
display(df.head())
```

```python
for colunas in df.columns:
    print(colunas)
```

```python
# Qual o produto mais vendido?
maisVendido = round(df.groupby('Produto')['Quantidade'].sum().sort_values(ascending=False).head(1))
print(maisVendido)
```

```python
# Relacione as vendas por fornecedor.
vendForn = df.groupby('Fornecedor')['ValorUnitário'].sum().sort_index()
print(vendForn)
```

```python
# Data que os produtos foram mais vendidos.
data = round(df.groupby('Data')['Quantidade'].sum().sort_values(ascending=False).head(1))
print(data)
```
