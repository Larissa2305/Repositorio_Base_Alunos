### Testar visualizações com dois datasets do seaborn

```python
import matplotlib as plt
import seaborn as sns

titanic = sns.load_dataset('titanic')
#display(titanic)

for i in titanic.columns:
    print(i)
```

```python
sns.barplot(
    data=titanic,
    x='embarked',
    y='class',
    hue='parch',
)
```

```python
sns.histplot(
    data=titanic,
    x='age',
    hue='class',
    kde=True
)
```
