# Aula 07 - Gráficos com Matplotlib

```python
import matplotlib.pyplot as plt

copasBrasil = {
    1958: "Suécia",
    1962: "Chile",
    1970: "México",
    1994: "Estados Unidos",
    2002: "Coreia/Japão",
}

anos = copasBrasil.keys()
pais = copasBrasil.values()

plt.figure(figsize=(6,3), facecolor='#fedd00')
plt.title('Anos em que o Brasil foi campeão mundial')
plt.plot(pais, anos, color='#009739') # gráfico de linha, país = x, anos = y
plt.xlabel('País Sede')
plt.ylabel('Ano')

plt.show()
```

## Gráfico de colunas

```python
meses = ['Janeiro','Fevereiro','Março','Abril','Maio','Junho']
vendas = [150,285,682,450,500,550]

plt.figure(figsize=(7,4))
plt.bar(meses,vendas, color=['violet','pink'])
plt.title('quantidade de Vendas - Primeiro Semestre 2026')

plt.show()
```

```python
import pandas as pd

df = pd.read_excel('Base_vendas.xlsx')

df['Faturamento'] = df['Qtd Vendida'] * df['Preço Unitário']
display(df.head())

for i in df.columns:
    print(i)
```

```python
FaturamentoLoja =df.groupby('Loja')['Faturamento'].sum().sort_values(ascending=False)

print(FaturamentoLoja)

plt.figure(figsize=(15,4))
FaturamentoLoja.plot(kind='barh')


plt.show()


#print(FaturamentoLoja)
```

```python
# Um gráfico de pizza, apresentando a quantidade de produtos por gênero

QtdGenero = df.groupby('Gênero')['Qtd Vendida'].sum()
print(QtdGenero)

plt.figure(figsize=(7,4))
plt.title('Quantidade de Produtos por Gênero')
pizza = QtdGenero.plot(kind='pie', colors=['pink','violet'], autopct='%1.1f%%')
pizza.legend()

plt.show()
```
